## Common Python Exception Types

| Exception Type | Description | Example Code |
|---|---|---|
| `IndexError` | Happens when attempting to use an index that is not available in a list or sequence. | `my_list = [1, 2]; print(my_list[5])` |
| `ValueError` | Occurs when a value is not suitable for the operation being performed. | `int("hello")` |
| `KeyError` | Happens when a program tries to retrieve a dictionary entry using a key that is missing. | `my_dict = {"name": "John"}; print(my_dict["age"])` |
| `EOFError` | Occurs when `input()` expects more data but reaches the end of the input stream. | `name = input()` *(when input reaches EOF)* |
| `ZeroDivisionError` | Happens when a division or modulo operation uses zero as the divisor. | `print(10 / 0)` |

## The try...except Statement

To prevent an exception from crashing your program, you wrap the risky code inside a try block. If an error occurs, Python immediately jumps to the corresponding except block instead of crashing.

```python
# Safe user input conversion
try:
    user_input = input("Enter an integer: ")
    number = int(user_input)
    print(f"Success! Your number squared is {number ** 2}")
except:
    print("Error: You must enter a valid integer.")
```

## Multiple exception handlers

A single try block can raise different error types (such as ValueError, NameError, or TypeError). To handle each unique scenario, you can add multiple except blocks by specifying the exact exception type for each clause.  
The `as` keyword binds a name to the exception being handled.

```python
# Example: Handling multiple different exception types
try:
    user_input = input("Enter a whole number to divide 50 by: ")
    number = int(user_input)
    result = 50 / number
    print(f"The result is: {result}")

# Handles non-integer text input
except ValueError:
    print("Error: Please enter a valid integer, not text.")

# Handles division by zero
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")

# Catch-all fallback for any other unexpected runtime errors
except Exception as err:
    print(f"An unexpected error occurred: {err}")
```

## Manual Exception Raising

You can also explicitly trigger exceptions in your own code using the raise keyword when a condition violates your program's logic:

```python
raise ValueError                      # Raises a ValueError with no specific error message.
raise ValueError("number is prime")   # Raises a ValueError with the custom message "number is prime".
```

```python
# Manually raising exceptions with custom messages
def check_positive(number):
    if number < 0:
        raise ValueError("Number cannot be negative.")
    return f"Valid number: {number}"

try:
    print(check_positive(-5))
except ValueError as err:
    print(f"Caught an error: {err}")
```
