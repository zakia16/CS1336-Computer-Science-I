# Lists

Like a string, a list is a sequence of values. In a string, the values are characters; in a list, they can be any type. The values in a list are called elements.
There are several ways to create a new list; the simplest is to enclose the elements in square brackets (`[ and ]`). For example, here is a list of two integers.

````python
numbers = [42, 123]
````

And here’s a list of three strings.

````python
cheeses = ['Cheddar', 'Edam', 'Gouda']
````

The elements of a list don’t have to be the same type. The following list contains a string, a float, an integer, and even another list.

````python
my_list = ["foo", 1, "hello", [], None, 2.3]
````

A list within another list is nested.
A list that contains no elements is called an empty list; you can create one with empty brackets, [].

````python
empty_list = []
````

The len function returns the length of a list.

````python
len(my_list)
````

# List Membership


You can check if a specific element is in a list using the in operator. This operator returns True if the element is found in the list and False if it is not. Here’s an example:

````python
fruits = ["apple", "banana", "cherry", "date"]
check_fruit = "banana"

if check_fruit in fruits:
    print(check_fruit, "is in the list.")
else:
    print(check_fruit, "is not in the list.")
# output: banana is in the list.
````

# Exploring List Methods


````python
# 1. append() - Adds a single element to the end
my_list = [1, 2, 3]
my_list.append(4)
print("append:", my_list)  # [1, 2, 3, 4]

# 2. count() - Counts occurrences of a specific element
my_list = [1, 2, 2, 3, 2, 4]
print("count of 2s:", my_list.count(2))  # 3

# 3. copy() - Creates a shallow copy of a list
original_list = [1, 2, 3]
new_list = original_list.copy()
print("copy:", new_list)  # [1, 2, 3]

# 4. extend() - Appends all elements from another iterable/list
list1 = [1, 2, 3]
list2 = [4, 5, 6]
list1.extend(list2)
print("extend:", list1)  # [1, 2, 3, 4, 5, 6]

# Alternative: Concatenation using the + operator
result = [1, 2, 3] + [4, 5, 6]
print("concatenation (+):", result)  # [1, 2, 3, 4, 5, 6]

# 5. pop() - Removes and returns an element by its index
my_list = [1, 2, 3, 4]
popped_element = my_list.pop(2)
print("pop item:", popped_element)  # 3
print("pop list:", my_list)  # [1, 2, 4]

# 6. remove() - Removes the first occurrence of a value
my_list = [1, 2, 3, 2, 4]
my_list.remove(2)
print("remove:", my_list)  # [1, 3, 2, 4]


# list.sort() METHOD vs sorted() FUNCTION COMPARISON
#==========================================================
# 7. sort() - Modifies the list IN-PLACE and returns None
my_list = [3, 1, 4, 1, 5, 9, 2]
return_val = my_list.sort()
print("sort() modifies in-place:", my_list)     # [1, 1, 2, 3, 4, 5, 9]
print("sort() return value:", return_val)       # None

# 8. sorted() - Returns a NEW sorted list, leaving the original unchanged
scramble = ['c', 'a', 'b']
new_sorted_list = sorted(scramble)
print("sorted() returns new list:", new_sorted_list)  # ['a', 'b', 'c']
print("original remains unchanged:", scramble)        # ['c', 'a', 'b']

# sorted() works on any iterable (like strings) and can be combined with join()
sorted_string = ''.join(sorted('letters'))
print("sorted string with join():", sorted_string)  # 'eelrstt'

# 9. reverse() - Reverses the order of elements in-place
my_list = [1, 2, 3, 4, 5]
my_list.reverse()
print("reverse:", my_list)  # [5, 4, 3, 2, 1]
````

Program to manage and analyze student test scores using list methods

````python
# 1. Start with the initial final exam scores
final_exam_scores = [78.5, 85.0, 92.0, 64.5, 88.0]
print("Initial final exam scores:", final_exam_scores)

# 2. len() - Find out how many students have submitted their exams so far
num_students = len(final_exam_scores)
print(f"Total number of scores recorded: {num_students}")

# 3. + Operator (List Concatenation) - Combine the main list with a list of late submissions (though late sunmssion is not possible in CS I !!!)
late_submissions = [76.0, 95.5]
final_exam_scores = final_exam_scores + late_submissions
print("After combining lists using '+':", final_exam_scores)

# 4. sort() - Organize the scores in-place from lowest to highest for grading analysis
final_exam_scores.sort()
print("Sorted final exam scores (in-place):", final_exam_scores)

# 5. len() - Recheck the total class size after adding late submissions
total_class_size = len(final_exam_scores)
print(f"Updated total class size: {total_class_size}")

# 6. pop() - Remove and review the lowest score
lowest_score = final_exam_scores.pop(0)
print(f"Removed lowest score ({lowest_score}) for grade adjustment.")
print("Final active scores list:", final_exam_scores)
````

# List Slicing


````python
my_list = [1, 'hello', 22, 2.7, 'python']

# 1. Basic Slicing [start:end] (end index is exclusive)
print(my_list[1:3])     # Output: ['hello', 22]

# 2. Omitting Start or End Index
print(my_list[:3])      # From beginning to index 2: [1, 'hello', 22]
print(my_list[3:])      # From index 3 to the end: [2.7, 'python']

# 3. Creating a Full Copy
list_copy = my_list[:]  # Returns a separate copy of the list

# 4. Negative Indexing (counts from the end)
print(my_list[3:-1])    # From index 3 up to the element before the last: [2.7]
````

# Passing Lists to Functions

When you pass a list into a function, Python passes the original list, not a copy. This means if the function modifies the list (like using .append() or .sort()), it will permanently change your original list outside the function. This is called a side effect.
  
**The Problem:** If a function changes your list by accident, it can cause bugs in your program.

**The Solution:** To keep your original list safe, you can pass a copy of the list into the function using slicing ([:]) or the .copy() method.

````python
# Passing Lists to Functions: Side Effects vs. Protection
def modify(num_list):
    num_list[1] = 99  # Modifies the list passed into the function


# --- Scenario 1: Passing the original list directly ---
print("--- Scenario 1: Passing Original List ---")
my_list = [10, 20, 30]
modify(my_list)
print(my_list)  # Output: [10, 99, 30] (Original list was modified!)


# --- Scenario 2: Passing a copy using slicing ([:]) ---
print("\n--- Scenario 2: Passing a Copy ---")
my_list = [10, 20, 30]
modify(my_list[:])  # Pass a copy of the list
print(my_list)  # Output: [10, 20, 30] (Original list remains safe!)
````
