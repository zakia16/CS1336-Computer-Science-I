As our programs grow beyond a single script, putting everything into one file becomes hard to read, test, and maintain. Modules solve this by letting you split code into separate `.py` files with clear responsibilities, then connect them using `import`.

A Python module is a file containing Python definitions and statements. A module can define functions, classes, and variables. A module can also include runnable code.  
Modules are divided into three types.
  •	User-defined Modules  
  •	Predefined Modules(Built-in Modules)  
  •	External Modules  
**	Ture power of Python lies in its extensive libraries and modules.

# User-defined Modules

Python gives its programmers is the ability to create user-defined modules.
In Python, any .py file is a module. Its module name is the filename without the .py extension.

Suppose you have three files in the same folder:

main.py — coordinates input and calls functions from other modules

coffee_bar.py — handles logging customer orders

roast_batch.py — calculates and grinds bean

```python
# coffee_bar.py

# As I am running from goolge colab, Colab-only magic command to save this cell as a .py file (omit in local IDEs)
%%writefile coffee_bar.py
def place_order(patron_name, cup_count):
    print(f"Order logged for {patron_name}: {cup_count} cup(s) of coffee.")
```

```python
# roast_batch.py

# As I am running from goolge colab, Colab-only magic command to save this cell as a .py file (omit in local IDEs)
%%writefile roast_batch.py
def grind_beans(cup_count):
    grams = cup_count * 15
    print(f"Grinding {grams}g of fresh beans...")
```

```python
# main.py

import coffee_bar
import roast_batch

patron_name = input()
cup_count = int(input())

coffee_bar.place_order(patron_name, cup_count)
roast_batch.grind_beans(cup_count)
```

When you write import coffee_bar, Python loads coffee_bar.py and gives you access to its functions and variables using `dot notation`.
