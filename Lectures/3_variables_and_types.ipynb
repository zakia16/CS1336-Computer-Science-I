#  Integer Division and Modulus

When working with numbers in Python, standard division (/) always returns a floating-point number. However, two related operators allow you to work strictly with whole numbers and remainders: **integer division (//)** and **the modulus operator (%).**

## Integer Division (//)
The integer division operator, //, divides two numbers and rounds down to the nearest whole integer (also known as floor division).

````python
minutes = 105
minutes / 60
# Output: 1.75
````

But we don’t normally write hours with decimal points. Integer division returns the integer number of hours, rounding down:

````python
minutes = 105
hours = minutes // 60
hours
# Output: 1
````

## The Modulus Operator (%)
To find the leftover minutes after extracting the whole hours, you could calculate it manually:

````python
remainder = minutes - hours * 60
remainder
````

Or you could use the modulus operator, %, which divides two numbers and returns the remainder.

````python
remainder = minutes % 60
remainder
````

## Practical Applications of Modulus
The modulus operator is far more versatile than simple remainder calculations. Here are three common patterns in programming:

### 1. Checking Divisibility
If x % y evaluates to 0, then x is perfectly divisible by y. This is commonly used to test if a number is even or odd:

````python
x % 2 == 0  # Evaluates to True if x is even
````

### 2. Extracting Digits

Modulo arithmetic in base 10 allows you to isolate the rightmost digits of any integer:   

x % 10 extracts the last digit.   
x % 100 extracts the last two digits.

````python
x = 123

print (x % 10)
# Output: 3

print(x % 100)
# Output: 23
````

### 3. Modular ("Clock") Arithmetic
Modulus naturally constrains numbers to a repeating cycle or loop. For example, calculating time on a 12-hour clock wraps back around after reaching 12:

````python
start = 11      # 11 AM
duration = 3    # 3 hours long

end = (start + duration) % 12
end
# Output: 2 (The event ends at 2 PM)
````

There are many more uses for these operators, which we will return to in later chapters.

# Type Conversion/ Type Casting

Type conversion, also known as type casting, is the process of converting a value from one data type to another. Python provides several built-in functions to perform type conversion, which can be categorized into two types: **implicit** and **explicit**.

**1. Implicit Type Conversion:**

  Implicit type conversion happens automatically when Python converts one data type to another without any explicit instruction from the user. This usually occurs when combining different data types in an operation.


````python
# Example

# Adding an integer and a float
integer_value = 10
float_value = 5.5

# Python automatically converts the integer to float before addition
result = integer_value + float_value
print(result)  # Output: 15.5
print(type(result))  # Output: <class 'float'>
````

**2. Explicit Type Conversion**

Explicit type conversion, or type casting, is when the programmer manually converts one data type to another using built-in functions. This type of conversion is necessary when you need to control the conversion process and ensure that it happens as intended.

| Target Type | Function | Valid Inputs & Examples | What It Does |
| :--- | :--- | :--- | :--- |
| **Integer** | `int(x)` | Whole numbers, floats, string digits<br><br>`int(3.9)` $\rightarrow$ `3`<br>`int("42")` $\rightarrow$ `42` | Drops decimals (does not round). Errors on letters or symbols. |
| **Float** | `float(x)` | Integers, decimal strings<br><br>`float(5)` $\rightarrow$ `5.0`<br>`float("3.14")` $\rightarrow$ `3.14` | Converts numbers to decimals. Does not work on fraction strings like `"1/2"`. |
| **String** | `str(x)` | Any Python value<br><br>`str(100)` $\rightarrow$ `"100"`<br>`str([1, 2])` $\rightarrow$ `"[1, 2]"` | Turns any object into readable text. Works universally. |

#Primitive types
For primitive types, we can just use their type names. The table below gives the names of the common primitive types that are built into Python. There are other built-in types that are omitted because we tend not to use them in this course.

| Python Type | Description | Example Literals |
| :--- | :--- | :--- |
| `int` | Whole numbers (positive, negative, or zero) | `42`, `-15`, `0` |
| `float` | Real numbers containing a decimal point | `3.14`, `-0.001`, `10.0` |
| `str` | Text sequence enclosed in single or double quotes | `'python'`, `"hello"`, `""` |
| `bool` | Truth values representing boolean logic | `True`, `False` |

#Compound types
For compound types like lists, dictionaries, and tuples, we can also just use their type names: list, dict, and tuple.But often we need to be more specific. For example, often we want to say that a function takes in not just any list, but only a list of integers; we might also want to say that this function returns not just any tuple, but a tuple containing one string and one boolean value.

We can use square brackets to express the “contained” types for each of these compound types. The table below shows examples of this syntax for each type; the capitalized words in square brackets could be substituted with any type.

| Data Type | Description  | Example |
| :--- | :--- | :--- |
| `list[T]` | An ordered, mutable collection where all contained elements match type `T` | `[10, 20, 30]` has type `list[int]` |
| `dict[KeyType, ValType]` | Key-value pairs mapping unique keys of `KeyType` to values of `ValType` | `{"age": 25, "score": 98}` has type `dict[str, int]` |
| `tuple[T1, T2, ...]` | An immutable, ordered sequence with fixed positions corresponding to specified element types | `("Alice", 95.5, True)` has type `tuple[str, float, bool]` |

````python
# Checking data types

print(type(10))                  # Int
print(type(3.14))                # Float
print(type(1 + 3j))              # Complex
print(type('Alice'))             # String
print(type([1, 2, 3]))           # List
print(type({'name': 'Alice'}))   # Dictionary
print(type({9.8, 3.14, 2.7}))    # Tuple
````

Each character can be accessed by its index either by forward indexing or by backward indexing.

**Index:** a number specifying the position of an element in a string/list

image.png

````python
# @title Default title text
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

def visualize_string_indexing(text):
    if not text:
        print("Please enter a non-empty string.")
        return

    n = len(text)
    indices = list(range(n))
    neg_indices = [i - n for i in indices]
    characters = list(text)

    # Scale figure size dynamically
    fig, ax = plt.subplots(figsize=(max(6, n * 1.2), 2.5))

    # Background match (Sage Green canvas)
    fig.patch.set_facecolor('#88C070')
    ax.set_facecolor('#88C070')

    box_style = "round,pad=0.02,rounding_size=0.08"

    for i, (char, pos_i, neg_i) in enumerate(zip(characters, indices, neg_indices)):
        # Top Box: Positive Index (Darker Translucent Green accent)
        ax.add_patch(FancyBboxPatch((i * 1.1, 1.2), 1.0, 0.7, boxstyle=box_style, color='#5E9447', ec='#335223', lw=1.5))
        ax.text(i * 1.1 + 0.5, 1.55, f"[{pos_i}]", fontsize=12, ha='center', va='center', color='#FFFFFF', weight='bold')

        # Middle Box: Character Value (Lighter Mint Green box)
        ax.add_patch(FancyBboxPatch((i * 1.1, 0.4), 1.0, 0.7, boxstyle=box_style, color='#A9DC90', ec='#426932', lw=1.5))
        ax.text(i * 1.1 + 0.5, 0.75, repr(char), fontsize=14, ha='center', va='center', color='#1A330E', weight='bold')

        # Bottom Box: Negative Index (Darker Olive/Forest Green accent)
        ax.add_patch(FancyBboxPatch((i * 1.1, -0.4), 1.0, 0.7, boxstyle=box_style, color='#3B612C', ec='#1E3614', lw=1.5))
        ax.text(i * 1.1 + 0.5, -0.05, f"[{neg_i}]", fontsize=12, ha='center', va='center', color='#E2F4D8', weight='bold')

    # Formatting plot limits
    ax.set_xlim(-0.2, n * 1.1 + 0.1)
    ax.set_ylim(-0.6, 2.1)
    ax.axis('off')

    plt.title(f"String Index Structure for: {repr(text)}", loc='center', pad=15, fontsize=13, color='#1A330E', weight='bold')
    plt.tight_layout()
    plt.show()

# Example usage
user_word = input("Enter any string to visualize: ")
visualize_string_indexing(user_word)
````
