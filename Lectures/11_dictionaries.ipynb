A dictionary in Python is a container object including key-value pairs.While lists and tuples use integer indices to order elements, dictionaries establish a mapping from unique keys to values.

Just like a physical phone book maps a Name to a Phone Number, a Python dictionary maps a unique Key to a Value.

### Real-World vs. Python Dictionary

Just like a physical phone book maps a **Name** to a **Phone Number**, a Python dictionary maps a unique **Key** to a **Value**.

| Real-World Phone Book | Python Dictionary |
| :--- | :--- |
| **Name** (e.g., `"Mom"`) | **Key** (e.g., `"Sarah"`) |
| **Phone Number** (e.g., `"555-0192"`) | **Value** (e.g., `"555-8321"`) |

### Creating and Accessing Dictionaries

You can initialize an empty dictionary using empty `curly braces {}` or the` dict()` constructor.   
Items are added and looked up using bracket notation:

````python
# Create an empty dictionary and add mappings
numbers = {}
numbers['zero'] = 0
numbers['one'] = 1
numbers['two'] = 2

print(numbers)              # Output: {'zero': 0, 'one': 1, 'two': 2}

# Accessing a key
print(numbers['two'])       # Output: 2

# Accessing a key that does not exist raises a KeyError
# print(numbers['three'])   # Raises KeyError: 'three'

# Modifying an existing key
numbers['two'] = 22
print(numbers)
````

You can also initialize a dictionary all at once by separating keys and values with **colons**:

````python
numbers = {'zero': 0, 'one': 1, 'two': 2}
print(numbers)
````

### Common Dictionary Operations

In Python, a dictionary is a mutable data type, which means that a dictionary's content can be modified after creation. Dictionary items can be added, updated, or deleted from a dictionary after a dictionary object is created.

| Operation / Method | Description | Example |
|---|---|---|
| `len(d)` | Returns the number of key-value pairs in `d`. | `len({'a': 1, 'b': 2}) → 2` |
| `d.update(other)` | Merges key-value pairs from `other` into `d`. | `d.update({'c': 3})` |
| `d.pop(key)` | Removes a key and returns its value, or raises `KeyError` if the key doesn't exist. | `d.pop('a')` |

````python
# Initialize a sample dictionary
d = {'a': 1, 'b': 2}

# 1. len(d): Returns the number of key-value pairs
print(len(d))                 # Output: 2

# 2. d.update(other): Merges key-value pairs from another dictionary
d.update({'c': 3})
print(d)                      # Output: {'a': 1, 'b': 2, 'c': 3}

# 3. d.pop(key): Removes the key and returns its value
val = d.pop('a')
print(val)                    # Output: 1
print(d)                      # Output: {'b': 2, 'c': 3}
````

### Conditionals and looping in dictionaries

Conditional statements can be used with dictionaries to check keys or evaluate stored values using direct lookups:

````python
# Checking if a key exists: Use the in operator directly on the dictionary

movies = {"The godfather": 1974, "Interstellar": 2014}
print("Interstellar" in movies)  # Output: True
````

````python
# Evaluating a value: Access the dictionary value via bracket notation and compare it using standard operators (<, >, ==)

movies = {"The godfather": 1974, "Interstellar": 2014}
print(movies["The godfather"] < 2000)  # Output: True
````

````python
# code to check if an item exists in a dictionary

word_dict = {"cat": 2, "dog": 1}
item = "cat"
if item in word_dict:
    print("Found!")
else:
    print("Not found!")
````

### Looping on a Dictionary

A for loop over a dictionary automatically iterates through its keys. You can use each key to access its corresponding value via **bracket** notation:

````python
zip_codes = {"Berkeley": 94709, "Santa Cruz": 95064, "Mountain View": 94030}

for key in zip_codes:
    print(key, zip_codes[key])
````

### Dictionary Lookup Efficiency and Hash Tables

Dictionary items are stored in a hash table, meaning the `in` operator takes about the same amount of time to find a key regardless of how many items are in the dictionary.

To see this performance difference, we can compare two approaches for finding reverse-word pairs (like stressed and desserts). First, we load our word collection into a list from an external text file containing over 113,000 words:

````python
# Load all words from the external file into a list
word_list = open("words.txt").read().split()  # returns a list: ['cat', 'dog', 'bird', ........upto 113,000 words]
````

**The Slow Approach (List Search):**  
Searching a list requires scanning elements sequentially from the beginning. Checking over 100,000 words inside a loop results in billions of comparisons, taking over a minute to run.

````python
def too_slow(word_list):
    count = 0
    for word in word_list:
        if reverse_word(word) in word_list:
            count += 1
    return count
````

**The Fast Approach (Dictionary Hash Lookup):**
Storing words as keys in a dictionary enables constant-time lookups. This version runs in less than one-hundredth of a second. Its about 10,000 times faster than the list version.

````python
word_dict = {}
for word in word_list:
    word_dict[word] = 1

def much_faster(word_dict):
    count = 0
    for word in word_dict:
        if reverse_word(word) in word_dict:
            count += 1
    return count
````

## Nested dictionaries

Python dictionaries are a type of data structure that allows for storing data in **key-value** pairs. Nested dictionaries are dictionaries that are stored as values within another dictionary. Ex: An organizational chart with keys being different departments and values being dictionaries of employees in a given department. For storing employee information in a department, a dictionary can be used with keys being employee IDs and values being employee names. The code below shows the structure of such nested dictionaries and how nested values can be accessed.

````python
# Defining nested dictionaries
company_org_chart = {
  "Marketing": {
    "ID234": "Jane Smith"
  },
  "Sales": {
    "ID123": "Bob Johnson",
    "ID122": "David Lee"
  },
  "Engineering": {
    "ID303": "Radhika Potlapally",
    "ID321": "Maryam Samimi"
  }
}
````

````python
# Accessing nested dictionary items
print(company_org_chart["Sales"]["ID122"])
print(company_org_chart["Engineering"]["ID321"])
````
