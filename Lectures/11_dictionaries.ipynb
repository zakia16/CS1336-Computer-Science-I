A dictionary in Python is a container object including key-value pairs.While lists and tuples use integer indices to order elements, dictionaries establish a mapping from unique keys to values.

### Creating and Accessing Dictionaries

You can initialize an empty dictionary using empty curly braces {} or the dict() constructor.   
Items are added and looked up using bracket notation:

```python
# Create an empty dictionary and add mappings
numbers = {}
numbers['zero'] = 0
numbers['one'] = 1
numbers['two'] = 2

print(numbers)              # Output: {'zero': 0, 'one': 1, 'two': 2}

# Accessing a key
print(numbers['two'])       # Output: 2

# Accessing a key that does not exist raises a KeyError
# print(numbers['three'])   # Raises KeyError: 'three'

# Modifying an existing key
numbers['two'] = 22
print(numbers)
```

You can also initialize a dictionary all at once by separating keys and values with colons:

```python
numbers = {'zero': 0, 'one': 1, 'two': 2}
print(numbers)
```

### Common Dictionary Operations

In Python, a dictionary is a mutable data type, which means that a dictionary's content can be modified after creation. Dictionary items can be added, updated, or deleted from a dictionary after a dictionary object is created.

| Operation / Method | Description | Example |
|---|---|---|
| `len(d)` | Returns the number of key-value pairs in `d`. | `len({'a': 1, 'b': 2}) → 2` |
| `key in d` | Checks if a key exists in `d` (runs in $O(1)$ time). | `'a' in {'a': 1} → True` |
| `d.update(other)` | Merges key-value pairs from `other` into `d`. | `d.update({'c': 3})` |
| `d.pop(key)` | Removes a key and returns its value, or raises `KeyError` if the key doesn't exist. | `d.pop('a')` |

```python
# Initialize a sample dictionary
d = {'a': 1, 'b': 2}

# 1. len(d): Returns the number of key-value pairs
print(len(d))                 # Output: 2

# 2. key in d: Checks if a key exists in the dictionary
print('a' in d)               # Output: True
print('c' in d)               # Output: False

# 3. d.update(other): Merges key-value pairs from another dictionary
d.update({'c': 3})
print(d)                      # Output: {'a': 1, 'b': 2, 'c': 3}

# 4. d.pop(key): Removes the key and returns its value
val = d.pop('a')
print(val)                    # Output: 1
print(d)                      # Output: {'b': 2, 'c': 3}
```
