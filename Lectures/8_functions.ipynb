#  Functions

Python provides built-in functions like `int()` and `float()`, as well as functions from modules like `math.sqrt()` and `math.pow()`. In this chapter, you will learn to **define and call your own functions**, pass arguments, and call one function from another. Functions are fundamental to programming because they make code reusable and organized.

## 1. Defining Functions

A function definition uses def to define a function. The header ends with a colon (:), and the indented body contains the statements executed when the function is called.

```python
count = 1

if count <= 5:
    print("Hello World")
```

* def: Defines a function.   
* print_lyrics: Function name.
* (): Indicates no parameters.
* Function call: print_lyrics() executes the function.

## 2. Parameters and Arguments

```python
#------------------------------------- Method 1  "Condition-Controlled Loop"--------------------
count = 1
while count <= 5:  # Loop expression
    print("Hello")
    count = count + 1 # 6
```

##continue Statements

•	Skips the current iteration and moves to the next one.
•	The loop does not stop but continues with the next iteration.

```python
for num in range(1, 6):
    if num == 3:
        continue  # Skips the rest of the loop 			   # when num is 3
    print(num)
```

**The while loop is a conditional pretest loop**   
*   Iterates as long as a certain condition exits  
*	Validating input   
*	Reading lists of data terminated by a sentinel

**The for loop is a pretest loop**
*   Built-in expressions for initializing, testing, and updating
*   Situations where the exact number of iterations is known
