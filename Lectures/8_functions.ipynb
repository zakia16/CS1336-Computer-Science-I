#  Functions

Python provides built-in functions like `int()` and `float()`, as well as functions from modules like `math.sqrt()` and `math.pow()`. In this chapter, you will learn to **define and call your own functions**, pass arguments, and call one function from another. Functions are fundamental to programming because they make code reusable and organized.

## 1. Defining and Calling Functions

A function definition uses def to define a function. The header ends with a colon (:), and the indented body contains the statements executed when the function is called.

```python
# Define the function
def debug_code():
    print("I fixed one bug...")
    print("Now I have 10 new bugs!")

# Call the function
debug_code()
```

* def: Defines a function.   
* print_lyrics: Function name.
* (): Indicates no parameters.
* Function call: print_lyrics() executes the function.

## 2. Using Parameters and Arguments

A parameter is a variable in a function definition. An argument is the value passed when calling the function.

```python
# Define a function that accepts a parameter 'message' and prints it
def print_twice(message):
  print(message)
  print(message)

# Test it with different arguments
print_twice("Hello")
print_twice("Bye")
```

You can also use variables as arguments:

```python
def print_twice(message):
  print(message)
  print(message)

# Call the function
message = "Hello"
print_twice(message)
```

Multiple Parameters:
A function can have multiple parameters.

```python
def add(a, b):
    print("Sum:", a + b)

add(5, 3)
add(10, 7)
```

##3. Function Calling Functions

Functions can call other functions, allowing code to be organized into reusable tasks.

```python
def move(action, count):
    print(action * count)

def robot_verse():
    sound = "Beep! "
    move(sound, 3) # Function Calling Functions
    print("🤖 [Robot Spin!]")
    move(sound, 2) # Function Calling Functions

def robot_dance_party():
    print("--- Dance Routine Start ---")
    robot_verse() # Function Calling Functions
    print("--- Encore! ---")
    robot_verse() # Function Calling Functions

# Call the function
robot_dance_party()
```

Function Composition: robot_dance_party() calls robot_verse(), which calls the utility function move(), showing how smaller tasks chain together to build a larger routine.

String Multiplication: action * count efficiently repeats the sound string without needing a loop every single time.

## 4. Functions That Return Values

A function can use the return statement to send a result back to the code that called it.

```python
def add(a, b):
    return a + b

result = add(5, 3)
print(result)
```

The expression a + b is evaluated, and its result is returned. The returned value can be stored in a variable, printed, or used in another calculation.
