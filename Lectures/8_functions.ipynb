#  Functions

Python provides built-in functions like `int()` and `float()`, as well as functions from modules like `math.sqrt()` and `math.pow()`. In this chapter, you will learn to **define and call your own functions**, pass arguments, and call one function from another. Functions are fundamental to programming because they make code reusable and organized.

## 1. Defining and Calling Functions

A function definition uses def to define a function. The header ends with a colon (:), and the indented body contains the statements executed when the function is called.

```python
# Define the function
def debug_code():
    print("I fixed one bug...")
    print("Now I have 10 new bugs!")

# Call the function
debug_code()
```

* def: Defines a function.   
* print_lyrics: Function name.
* (): Indicates no parameters.
* Function call: print_lyrics() executes the function.

## 2. Using Parameters and Arguments

A parameter is a variable in a function definition. An argument is the value passed when calling the function.

```python
# Define a function that accepts a parameter 'message' and prints it
def print_twice(message):
  print(message)
  print(message)

# Test it with different arguments
print_twice("Hello")
print_twice("Bye")
```

You can also use variables as arguments:

```python
def print_twice(message):
  print(message)
  print(message)

# Call the function
message = "Hello"
print_twice(message)
```

Multiple Parameters:
A function can have multiple parameters.

```python
def add(a, b):
    print("Sum:", a + b)

add(5, 3)
add(10, 7)
```

## 4. Functions That Return Values

A function can use the return statement to send a result back to the code that called it.

```python
def add(a, b):
    return a + b

result = add(5, 3)
print(result)
```

The expression a + b is evaluated, and its result is returned. The returned value can be stored in a variable, printed, or used in another calculation.

## 5. Return Values and Conditionals

When using return inside conditional statements, ensure that every possible path through the program hits a `return` statement, otherwise the function will return `None`.

Consider this function:

```python
def check_number(x):
    if x > 0:
        return "Positive"

print(check_number(5))
print(check_number(-3))
```

* check_number(5): The condition is true, so the function returns `"Positive"`.

* check_number(-3): The condition is false. There is no else statement and no other return, so Python automatically returns` None`.

To fix it, add an else:

```python
def absolute_value(x):
    if x < 0:
        return -x
    else:
        return x

print(check_number(5))
print(check_number(-3))
```

### Dead Code
Dead code is code that never executes because the function has already returned.

```python
def greet():
    return "Hello"
    print("Goodbye")

print(greet())
```

The print("Goodbye") statement never runs because return "Hello" ends the function immediately.  
**Remember:** Once Python executes return, the function stops running.

## 6. Local Variables and Scope

A local variable is created inside a function and can generally be accessed only within that function.

```python
def calculate_square(number):
    result = number * number
    print(result)

calculate_square(5)
print(result) # NameError: name 'result' is not defined
```

The variable result is local to calculate_square(). Attempting to access result outside the function causes a NameError.

Local variables help prevent different parts of a program from accidentally interfering with each other.

##7. Function Calling Functions

Functions can call other functions, allowing code to be organized into reusable tasks.

```python
def move(action, count):
    print(action * count)

def robot_verse():
    sound = "Beep! "
    move(sound, 3) # Function Calling Functions
    print("🤖 [Robot Spin!]")
    move(sound, 2) # Function Calling Functions

def robot_dance_party():
    print("--- Dance Routine Start ---")
    robot_verse() # Function Calling Functions
    print("--- Encore! ---")
    robot_verse() # Function Calling Functions

# Call the function
robot_dance_party()
```

Function Composition: robot_dance_party() calls robot_verse(), which calls the utility function move(), showing how smaller tasks chain together to build a larger routine.

String Multiplication: action * count efficiently repeats the sound string without needing a loop every single time.

## 8. Incremental Development

Incremental development means building a program one small step at a time, testing each step, and fixing errors before continuing.

### Step 1: Start with a Stub

First, create the skeleton of the function with a placeholder return value to ensure your code runs and connects properly:

```python
def calculate_total(price, tax_rate):
    return 0.0  # Stub

print(calculate_total(100.0, 0.08))
# Output: 0.0
```

### Step 2: Add Scaffolding (Intermediate Prints)
Next, write out your logic step-by-step, adding temporary print() statements to inspect the intermediate values and catch math or logic bugs early:

```python
def calculate_total(price, tax_rate):
    tax_amount = price * tax_rate
    print("Debug - tax_amount:", tax_amount)  # Scaffolding

    total = price + tax_amount
    print("Debug - total before return:", total) # Scaffolding

    return total

print("Final:", calculate_total(100.0, 0.08))
# Output:
# Debug - tax_amount: 8.0
# Debug - total before return: 108.0
# Final: 108.0
```

### Step 3: Clean Up
Once you verify that the calculations are entirely correct, delete the temporary scaffolding print statements for clean, production-ready code:

```python
def calculate_total(price, tax_rate):
    tax_amount = price * tax_rate
    return price + tax_amount

print("Final:", calculate_total(100.0, 0.08))
# Output: Final: 108.0
```

## 7.Functions with Default Parameters

A default parameter has a value that Python uses when the caller does not provide an argument for that parameter.

```python
def greet(name="Guest"):
    print("Hello,", name)

greet()
greet("Sam")
```

When greet() is called without an argument, Python uses "Guest". When greet("Sam") is called, Python uses "Sam" instead.



## 8. Functions That Return Multiple Values

Python allows a function to return multiple values. The returned values can be assigned to separate variables.

```python
def calculate(a, b):
    return a + b, a - b

sum_result, difference = calculate(10, 4)

print("Sum:", sum_result)
print("Difference:", difference)
```

Python returns the two values as a tuple, which can be unpacked into separate variables.
