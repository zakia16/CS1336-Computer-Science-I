# LOOPS


  A loop is a program construct that repeatedly executes the loop's statements (known as the loop body) while the loop's expression is true; when false, execution proceeds past the loop. Each time through a loop's statements is called an `iteration`.     
  In Python, we mainly have three types of loops:    
  for, while, and do-while.

#  while Loop Statements

Let’s look at an **if statement** and a **while loop** that use the same condition and take the same actions based on that condition.      
Here is the code with an **if** statement:

````python
count = 1

if count <= 2:
    print("Hello World")
````

Here is the code with a while statement:

````python
#------------------------------------- Method 1  "Condition-Controlled Loop"--------------------
count = 1                 # 1. Initialization
while count <= 2:         # 2. Condition
    print("Hello")        #    Body
    count = count + 1     # 3. Increment / Update
````

##While Loop using Sentinel value

A sentinel value is a special value used to signal the termination of a loop. In the example below, the loop continues until the user enters 'exit', which acts as the sentinel value to stop execution.

````python
user_input = input()

while user_input != 'exit':
	print(f"You entered: {user_input}")
	user_input = input("Type 'exit' to quit): ")

print("Exiting the program.")
````

````python
import io
from contextlib import redirect_stdout  # <-- Correct library


def run_zybook_tests(student_code_func, test_cases):
  """Executes student code against test cases and reports passes/failures."""
  passed = 0
  total = len(test_cases)

  print(f"{'='*15} TEST RESULTS ({total} Test Cases) {'='*15}\n")

  for i, (input_val, expected_output) in enumerate(test_cases, 1):
    buffer = io.StringIO()
    try:
      with redirect_stdout(buffer):
        student_code_func(input_val)
      actual_output = buffer.getvalue()

      if actual_output == expected_output:
        print(f"✅ Test Case {i}: PASSED")
        print(f"   Input:    {input_val}")
        print(f"   Output:   {repr(actual_output)}\n")
        passed += 1
      else:
        print(f"❌ Test Case {i}: FAILED")
        print(f"   Input:    {input_val}")
        print(f"   Expected: {repr(expected_output)}")
        print(f"   Actual:   {repr(actual_output)}\n")

    except Exception as e:
      print(f"💥 Test Case {i}: ERROR DURING EXECUTION")
      print(f"   Input: {input_val}")
      print(f"   Error: {e}\n")

  score = (passed / total) * 100
  print(f"{'='*45}")
  print(f"SCORE: {passed}/{total} ({score:.0f}%)")
````

````python
# ==========================================
# STUDENT SUBMISSION AREA
# ==========================================
def student_solution(user_val):
  # Student writes their code here
  while user_val > 0:
    print(user_val % 3, end="")
    user_val = user_val // 10
  print()


# ==========================================
# ZYBOOKS-STYLE TEST CASES
# ==========================================
# Format: (input_value, expected_output_string)
test_cases = [
    (927567, "765729\n"),  # Standard multi-digit number
    (12345, "54321\n"),  # Sequential digits
    (8000, "0008\n"),  # Trailing zeros
    (7, "7\n"),  # Single digit
    (1001, "1001\n"),  # Palindromic number
]

# Run the grader
run_zybook_tests(student_solution, test_cases)
````
