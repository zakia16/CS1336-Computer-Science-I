Up to this point, all the data in your programs has lived temporarily in computer memory (RAM); once your program finishes running, that data disappears. To save data permanently, you need to work with files stored on your computer's file system.

Three steps when a program uses a file:  
1.    Open the file  
2.    Process the file  
3.    Close the file  

## Reading from files

### 1. Open the file:
Reading information from and writing information to files is a common task in programming.

Python supports the opening of a file using the `open()` function.

### 2. Process the file
Python provides functions that can be called on a file object for reading the contents of a file:

The `read()` function reads the contents of a file and returns a string.   
The `readline()` function reads the next line in a file and returns a string.   
The `readlines()` function reads the individual lines of a file and returns a string list containing all the lines of the file in order.

### 3. Close the file
The `close()` method closes an open file.

You should always close your files, in some cases, due to buffering, changes made to a file may not show until you close the file.

Example: A file called mydata.txt has the following contents:

    Val magenta
    Zoe green
    Noa ochre
    Ina gray



```python
    """Demonstrating read() and readlines()"""

    # ------------ Using read()-------------------
    # Open the file and associate with a file object
    f1 = open("mydata.txt")

    # Read the contents of the file into a string
    contents = f1.read()

    # Print
    print(contents)

    # Always close the file once done using the file
    f1.close()

    #----------- Using readlines()-------------------
    # Open the file and associate with a file object
    f2 = open("input.txt")

    # Read the contents of the file into a string list
    contents = f2.readlines()

    # Print
    print(contents)

    # Always close the file once done using the file
    f2.close()
    """
    The code's output is:
    Result of using read():
    Val magenta
    Zoe green
    Noa ochre
    Ina gray

    Result of using readlines():
    ['Val magenta\n', 'Zoe green\n', 'Noa ochre\n', 'Ina gray']
    """
```

## Writing to files

### 1. Open the file:
A file may be opened for reading, allowing no changes, or for writing, allowing changes to occur in the file. The mode defines whether a file is opened for reading only or for writing.   
The default mode of the open() function is reading only. A second mode parameter defines the mode.

Ex: open("output.txt", 'w') opens the output.txt file in writing mode. The following table lists common modes.  
Please see details: https://openstax.org/books/introduction-python-programming/pages/14-2-writing-to-files

### 2. Process the file
The write() function is used to write to an already opened file. The write() function will only accept a string parameter. Other variable types must be cast to string before writing using write().

The write() function does not automatically add a newline character as the print() function does. A newline must be added explicitly by adding a newline ('\n') character.

### 3. Close the file
The write() function writes automatically to a temporary store called the file buffer. To ensure that the information is written to a file, the close() function must be used. The close() function finalizes changes and closes the file.

```python
    # Example: Writing to a file

    # Create a new file
    opfile = open("output.txt", 'w')

    # Writing to the file
    opfile.write("Writing to a new file.")

    # To add another line the newline character must be used
    opfile.write("\nSecond line.")

    # Ensure changes are saved by closing the file
    opfile.close()
```

## The 'with' statement


The preferred way of opening a file is with the with statement.   
Good practice is to use a with statement when opening files to guarantee that the file is closed when no longer needed.


```python
with open('mydata.txt', 'r') as f:  # Open file in read mode
    contents = f.readlines()

# File is automatically closed when 'with' block ends

print('\nContents of mydata.txt:')
print(contents)  # Print the list of lines

"""
Output:
Contents of mydata.txt:
['Val magenta\n', 'Zoe green\n', 'Noa ochre\n', 'Ina gray']
"""
```

We can use the with statement to open a file for reading and appending.

Example: A file called appendingfile.txt has the following contents:

    5
    7

```python
with open('appendingfile.txt', 'r+') as f:
    # Read in two integers
    num1 = int(f.readline().strip())
    num2 = int(f.readline().strip())

    product = num1 * num2

    f.write('\n')
    f.write(str(product))

# No need to call f.close() - f closed automatically
print('Closed myfile.txt')

"""
Output:
5
7
35
"""
```

### Another Approach: Using the Built-in csv Module

Because dealing with quoted fields, escaped characters, and custom delimiters is a common headache, Python provides a built-in module called csv. Instead of reinventing the wheel with .split(), we use csv.reader() to let Python handle all those edge cases safely and cleanly.

Example: A file called input.csv has the following contents:

    hello,cat,man,hey,dog,boy,
    Hello,man,cat,woman,dog,Cat,hey,boy

```python
import csv

with open("input.csv", 'r') as csvfile:
    user_file = csv.reader(csvfile, delimiter=',’)# Create a CSV reader object

    for row in user_file: # Read and print each row
        print(row)

"""
Output:
['hello', 'cat', 'man', 'hey', 'dog', 'boy', '']
['Hello', 'man', 'cat', 'woman', 'dog', 'Cat', 'hey', 'boy']
"""
```

## Interacting with a file system using the OS module

While core Python lets you open, read, and write files, it operates in a bit of a bubble. Your scripts don't inherently know how your computer's file system is structured, how to handle differences between operating systems, or how to manage folders programmatically.

The os module bridges that gap by solving key challenges:

*  Eliminating the "It Works on My Machine" Problem (Portability): Windows uses
backslashes (\\) while Mac/Linux use forward slashes (/). Hardcoding paths causes `FileNotFoundError` when shared. The os module abstracts these differences away.

*  Automating File System Maintenance: Scripts can dynamically create backup directories, organize files, or clean up temporary cache files automatically.

*  Safety Checks: You can verify whether a file or directory exists before trying to open it, preventing your program from crashing.

*  Accessing System Metadata: Functions like os.stat() let your scripts inspect low-level file attributes such as file size in bytes and last modified timestamps.

### Navigating and Checking Paths
Before reading or writing files, you often need to check your working environment or verify that a target path actually exists.

os.getcwd(): Returns the current working directory as a string.

os.path.exists(path): Returns True if the path exists, otherwise False.

os.path.isfile(path) / os.path.isdir(path): Verifies whether a path points specifically to a file or a directory.

```python
import os

# Check current working directory
current_dir = os.getcwd()
print(f"Current Directory: {current_dir}")

# Check if a file or folder exists
target = "logfile.txt"
if os.path.exists(target):
    if os.path.isfile(target):
        print(f"'{target}' exists and is a file.")
else:
    print(f"'{target}' does not exist.")
```

## Ensuring Portability with `os.path.join()`
A program's portability refers to its ability to run correctly across different operating systems without modification. Hardcoding file paths using OS-specific path separators leads to crashes when run on another machine.

To prevent these errors, always use the os.path module specifically os.path.join() which automatically concatenates path arguments using the correct separator for the host operating system:

```python
import os

# Avoid hardcoding OS-specific separators like: path = "subdir\\bat_mobile.jpg"
# Instead, use os.path.join for cross-platform portability:
path = os.path.join("subdir", "bat_mobile.jpg")
print(path)
# Result: "subdir\bat_mobile.jpg" on Windows, "subdir/bat_mobile.jpg" on Mac/Linux
```

# Creating and Managing Directories & Files

You can automate folder creation and file cleanup directly from your script—ideal for organizing logs, batch outputs, or datasets.

os.mkdir(path): Creates a single directory (raises an error if the parent directory is missing).

os.makedirs(path): Creates a directory and any necessary intermediate parent directories recursively.

os.remove(path): Deletes a file.

os.rmdir(path): Removes an empty directory.

```python
import os

# Create a new directory safely
dir_name = "logs_backup"
if not os.path.exists(dir_name):
    os.mkdir(dir_name)
    print(f"Created directory: {dir_name}")

# Create nested directories safely using exist_ok=True
os.makedirs("backups/2026/11", exist_ok=True)

# Remove a file if needed
# os.remove("temp_file.txt")
```
