# Multidimensional Lists in Python

Multidimensional lists in Python are lists that contain other lists as their elements. These nested lists create a structure similar to a grid or matrix, allowing you to work with more complex and structured data.

### Creating a Multidimensional List

```python
my_list = [[1, 2, 3], ["hello", "world"]]
```

### Accessing Lists within a Multidimensional List

To access the lists within a multidimensional list, use indexing. The first index selects a list from the outer list, and the second index selects an element within the inner list. For example:

```python
first_list = my_list[0]  # Select the first list
print(first_list)  # Output: [1, 2, 3]

second_list = my_list[1]  # Select the second list
print(second_list)  # Output: ["hello", "world"]
```

### Chaining Indices for Accessing Elements

To access specific elements within the inner lists, chain the indices. For instance:

```python
element = my_list[0][1]  # Access the element at the first index of the first list
print(element)  # Output: 2

word = my_list[1][0]  # Access the first element in the second list
print(word)  # Output: "hello"
```

### Using nested loops to iterate nested lists

A nested loop structure can be used to iterate a list-of-lists. For a two-dimensional list-of-lists, an outer for loop can be used for rows, and an inner for loop can be used for columns.

```python
"""Iterating a list-of-lists."""
# Create a list of numbers
list1 = [[1, 2, 3],
        [1, 4, 9],
        [1, 8, 27]]

# Iterating the list-of-lists
for row in list1:
  for cell in row:
    print(cell, end=" ")
  print()
```

```python
list1 = [[1, 2, 3],
        [1, 4, 9],
        [1, 8, 27]]

for i in range(len(list1)):
   # iterate through columns
    for j in range(len(list1[i])):
        print(list1[i][j], end=" ")
    print()
```

Nested lists are ideal for representing grid-based games like Tic-Tac-Toe. Each inner list represents a row on the board, and chained indexing allows you to check or update individual board coordinates.

```python
tic_tac_toe = [
    ["X", "O", "X"],
    [" ", "X", "X"],
    ["O", "O", ""]
]

# Accessing individual rows and elements
print(tic_tac_toe[0][0], tic_tac_toe[0][1], tic_tac_toe[0][2])
print(tic_tac_toe[1][0], tic_tac_toe[1][1], tic_tac_toe[1][2])
print(tic_tac_toe[2][0], tic_tac_toe[2][1], tic_tac_toe[2][2])
```

In the following example a Tic-Tac-Toe board is represented using a $3 \times 3$ nested list. First, the board is displayed by printing an underscore (_) for empty spaces (" ") to improve readability. Then, we simulate a turn where a player selects coordinates to make a move. The script checks whether the requested position is empty before updating the grid.

```python
tic_tac_toe = [
    ["X", "O", "X"],
    [" ", "X", "X"],
    ["O", "O", " "]
]

print("Current Board State:")
for row in tic_tac_toe:
    for cell in row:
        # Display an underscore for empty spots to make it look clean
        if cell == " ":
            print("_", end=" ")
        else:
            print(cell, end=" ")
    print()

# A player tries to make a move at the last row, last column (Row 2, Col 2)
row_choice = 2
col_choice = 2

print(f"\nPlayer attempts to play at Row {row_choice}, Column {col_choice}...")

# Check if the spot is empty (represented by an empty string "")
if tic_tac_toe[row_choice][col_choice] == "":
    tic_tac_toe[row_choice][col_choice] = "X"
    print("Move accepted! Winning placement locked in!")
else:
    print("🚨 Invalid move! Spot is already taken.")
```

# Tuples

Tuples are like lists. A tuple is a sequence of values. The values can be any type, and they are indexed by integers, so tuples are a lot like lists. The important difference is that tuples are immutable.


### Creating Tuples

To create a tuple, you can write a comma-separated list of values.

```python
t = 'l', 'u', 'p', 'i', 'n'
type(t)
```

Although it is not necessary, it is common to enclose tuples in parentheses.

```python
t = ('l', 'u', 'p', 'i', 'n')
type(t)
```

Single-element tuples: To create a tuple with a single element, you must include a trailing comma.

```python
# Single-element tuple (requires a trailing comma!)
t1 = ('p',)
print(type(t1))  # Output: <class 'tuple'>

# Without comma, Python treats it as a standard string in parentheses
t2 = ('p')
print(type(t2))  # Output: <class 'str'>
```

### Tuple Operators & Built-in Functions

Most list operators also work with tuples. For example:

```python
# 1. Create a base tuple
t = ('l', 'u', 'p', 'i', 'n')
print("Original tuple:", t)

# 2. Indexing & Slicing
first_char = t[0]             # Gives 'l'
middle_slice = t[1:3]         # Gives ('u', 'p')
print(f"Index 0: {first_char} | Slice [1:3]: {middle_slice}")

# 3. Concatenation (+) & Duplication (*)
combined = tuple('lup') + ('i', 'n')  # Concatenates into ('l', 'u', 'p', 'i', 'n')
duplicated = tuple('spam') * 2        # Duplicates into ('s', 'p', 'a', 'm', 's', 'p', 'a', 'm')
print("Concatenated:", combined)
print("Duplicated:", duplicated)

# 4. sorted() & reversed()
sorted_list = sorted(t)               # sorted() returns a list: ['i', 'l', 'n', 'p', 'u']
reversed_tuple = tuple(reversed(t))   # reversed() cast back to tuple: ('n', 'i', 'p', 'u', 'l')
print("Sorted (as list):", sorted_list)
print("Reversed (as tuple):", reversed_tuple)
```

### Mutability and performance

Tuples are immutable and have a fixed size, so tuples use less memory. Overall, tuples are faster to create and access, resulting in better performance that can be noticeable with large amounts of data.

# Sets

Python provides a class called set that represents a collection of unique elements.

## Creating sets

You can create a set in two primary ways:
1) Using Curly Braces {}

```python
# Duplicates like '88' will be automatically filtered out
final_grades = {85, 90, 88, 92, 88, 85}
print(final_grades)  # Output: {88, 85, 90, 92} (order may vary)
```

2. Using the set() Function

```python
empty_set = set()
```

You can pass any sequence (like a list or tuple) to set() to instantly remove duplicates:

```python
roster = ['Alice', 'Bob', 'Alice', 'Charlie']
unique_students = set(roster)  # {'Alice', 'Bob', 'Charlie'}
```

## Core Set Operations

```python
submissions_1 = {"Alice", "Bob", "Charlie", "Alice"}
submissions_2 = {"Charlie", "David", "Eve"}

# 1. Set Operations (Union, Intersection, Difference)
all_unique_submitters = submissions_1 | submissions_2    # Equivalent to: submissions_1.union(submissions_2)
print("All unique submitters:", all_unique_submitters)

both_sections = submissions_1 & submissions_2    # Equivalent to: submissions_1.intersection(submissions_2)
print("Students who submitted to both:", both_sections)

only_section_1 = submissions_1 - submissions_2   # Equivalent to: submissions_1.difference(submissions_2)
print("Only in Section 1:", only_section_1)

# 2. Adding and Discarding elements
active_roster = {"Alice", "Bob"}
active_roster.add("Frank")       # Add new student
active_roster.discard("Zack")    # Safe removal of non-existent student
print("Updated active roster:", active_roster)
```
