# Multidimensional Lists in Python

Multidimensional lists in Python are lists that contain other lists as their elements. These nested lists create a structure similar to a grid or matrix, allowing you to work with more complex and structured data.

### Creating a Multidimensional List

````python
my_list = [[1, 2, 3], ["hello", "world"]]
````

### Accessing Lists within a Multidimensional List

To access the lists within a multidimensional list, use indexing. The first index selects a list from the outer list, and the second index selects an element within the inner list. For example:

````python
first_list = my_list[0]  # Select the first list
print(first_list)  # Output: [1, 2, 3]

second_list = my_list[1]  # Select the second list
print(second_list)  # Output: ["hello", "world"]
````

### Chaining Indices for Accessing Elements

To access specific elements within the inner lists, chain the indices. For instance:

````python
element = my_list[0][1]  # Access the element at the first index of the first list
print(element)  # Output: 2

word = my_list[1][0]  # Access the first element in the second list
print(word)  # Output: "hello"
````

### Using nested loops to iterate nested lists

A nested loop structure can be used to iterate a list-of-lists. For a two-dimensional list-of-lists, an outer for loop can be used for rows, and an inner for loop can be used for columns.

````python
"""Iterating a list-of-lists."""
# Create a list of numbers
list1 = [[1, 2, 3],
        [1, 4, 9],
        [1, 8, 27]]

# Iterating the list-of-lists
for row in list1:
  for col in row:
    print(col, end=" ")
  print()
````

# Tuples

Tuples are like lists. A tuple is a sequence of values. The values can be any type, and they are indexed by integers, so tuples are a lot like lists. The important difference is that tuples are immutable.


### Creating Tuples

To create a tuple, you can write a comma-separated list of values.

````python
t = 'l', 'u', 'p', 'i', 'n'
type(t)
````

Although it is not necessary, it is common to enclose tuples in parentheses.

````python
t = ('l', 'u', 'p', 'i', 'n')
type(t)
````

Single-element tuples: To create a tuple with a single element, you must include a trailing comma. Without it, Python treats the parentheses as standard mathematical grouping:

````python
# Single-element tuple (requires a trailing comma!)
t1 = ('p',)
print(type(t1))  # Output: <class 'tuple'>

# Without comma, Python treats it as a standard string in parentheses
t2 = ('p')
print(type(t2))  # Output: <class 'str'>
````

### Tuple Operators & Built-in Functions

Most list operators also work with tuples. For example:

````python
# 1. Create a base tuple
t = ('l', 'u', 'p', 'i', 'n')
print("Original tuple:", t)

# 2. Indexing & Slicing
first_char = t[0]             # Gives 'l'
middle_slice = t[1:3]         # Gives ('u', 'p')
print(f"Index 0: {first_char} | Slice [1:3]: {middle_slice}")

# 3. Concatenation (+) & Duplication (*)
combined = tuple('lup') + ('i', 'n')  # Concatenates into ('l', 'u', 'p', 'i', 'n')
duplicated = tuple('spam') * 2        # Duplicates into ('s', 'p', 'a', 'm', 's', 'p', 'a', 'm')
print("Concatenated:", combined)
print("Duplicated:", duplicated)

# 4. sorted() & reversed()
sorted_list = sorted(t)               # sorted() returns a list: ['i', 'l', 'n', 'p', 'u']
reversed_tuple = tuple(reversed(t))   # reversed() cast back to tuple: ('n', 'i', 'p', 'u', 'l')
print("Sorted (as list):", sorted_list)
print("Reversed (as tuple):", reversed_tuple)
````
