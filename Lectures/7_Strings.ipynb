# String Methods

Python has many built-in methods that you can use to manipulate strings. We can create new strings from a string.

Because strings are immutable, Python string methods like `replace(old, new)` do not modify the original string in place. Instead, they create and return a new string object containing the updated content.

````python
var = ' Hello, world! '
print(var.replace('world', 'Python'))
````

More Methods:

````python
var = ' Hello, world! '
print(var.lower())  # Returns a copy of the string with all characters lowercased.
print(var.upper())  # Returns a copy of the string with all characters uppercased.
print(var.capitalize()) # Returns a copy of the string with the first character capitalized and the rest lowercased.
print(var.strip())  # Returns a copy of the string with leading and trailing whitespace removed.
````

Methods to check a string value that returns a True or False Boolean value.

````python
var = ' Hello, world! '
print(var.islower())  # Returns True if all cased characters are lowercase letters.
print(var.isupper())  # Returns True if all cased characters are uppercase letters.
print(var.isalpha())  # Returns True if the string consists only of letters and isn’t blank
print(var.isalnum())  # Returns True if the string consists only of letters and numbers (alphanumerics) and isn’t blank.
print(var.isdigit())  # Returns True if all characters are the numbers 0-9.
print(var.isspace())  # Returns True if all characters are whitespace.
print(var.startswith('Hello'))  # Returns True if the string starts with x.
print(var.endswith('world! '))  # Returns True if all characters are whitespace.
````

# Multiline Strings

 A multiline string in Python begins and ends with either three single quotes or three double quotes. Any quotes, tabs, or newlines in between the `“triple quotes”` are considered part of the string. Python’s indentation rules for blocks don’t apply to lines inside a multiline string.

````python
print('''Dear Alice,

Can you feed Eve's cat this weekend?

Sincerely,
Bob''')
````

 A multiline string is often used for comments.

# Strings are immutable

Strings in Python are immutable i.e, they cannot be modified in place. Attempting to reassign a character causes a TypeError:

````python
greeting = 'Hello, world!'
greeting[0] = 'h'
````

To modify a string:

````python
# use replace
greeting = 'Hello, world!'
greeting =   greeting.replace("H", "h") # 'hello, world!'
print(greeting)
````

   **OR**

````python
# create a new one using concatenation and slicing
greeting = 'Hello, world!'
new_greeting = 'h' + greeting[1:]  # 'hello, world!'
print(new_greeting)
````

  **OR**

````python
# Iterate over each character directly and replace any occurrences of 'H' with 'h'
greeting = 'Hello, world!'
new_greeting = ""

for char in greeting:
    if char == 'H':
        new_greeting += 'h'
    else:
        new_greeting += char

print(new_greeting)  # 'hello, world!'
````

However, using += inside a loop is not recommended for long strings. In-loop string concatenation (+=) re-allocates memory and re-copies all previous characters on every iteration, causing performance slowdowns.
